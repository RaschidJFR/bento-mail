## MailDev
Send emails to DevMail with curl:

```sh
curl \
--url 'smtp://localhost:1025' \
--mail-from randomuser@example.com \
--mail-rcpt mailserver@test.bento \
--upload-file - <<EOF                                                
$(cat 'samples/email/Canada Letter_ At the movies.eml')
EOF
```

````typescript
import MailDev from 'maildev'
process.env.NODE_TLS_REJECT_UNAUTHORIZED = "0";

const maildev = new MailDev();

maildev.listen();

maildev.on("new", (email) => {
  // We got a new email!
  console.log("New email received:", email);
});
````
